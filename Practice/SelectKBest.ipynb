```python
#import all libraries
# pandas- Handles data in table format
#train_test_split	Splits data into training and testing sets
#time	Measures execution time
#numpy	Performs numerical calculations
#StandardScaler	Scales numerical features
#SelectKBest	Selects the best features
#chi2	Measures the relationship between features and the target
#RFE	Removes less important features recursively
#LogisticRegression	Classification algorithm
#pickle	Saves and loads trained models
#matplotlib.pyplot	Creates graphs
import pandas as pd
from sklearn.model_selection import train_test_split 
import time
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.feature_selection import SelectKBest
from sklearn.feature_selection import chi2
from sklearn.feature_selection import RFE
from sklearn.linear_model import LogisticRegression
import pickle
import matplotlib.pyplot as plt

#Indep_X: Input features, such as age, salary, or experience.
#dep_Y: Target/output, such as selected or not selected.
#n: Number of features you want to select.
#chi2: Scores each feature based on its relationship with the target.
#fit(): Learns the feature scores.
#transform(): Keeps only the selected features.
#return: Sends the selected features back.
#Important: chi2 requires non-negative input values. If your features contain negative values, it can produce an error.
def selectkbest(Indep_X,dep_Y,n):
    test=SelectKBest(score_func=chi2,k=n)
    fit1=test.fit(Indep_X,dep_Y)
    selectk_features=fit1.transform(Indep_X)
    return selectk_features

#StandardScaler() converts features to a common scale, generally with mean 0 and standard deviation 1.
#fit_transform(X_train): Learns the training data's scaling values and scales the training data.
#transform(X_test): Uses the same scaling values to scale the test data.
def split_scalar(Indep_X,dep_Y):
    X_train,X_test, y_train,y_test=train_test_split(Indep_X,dep_Y,test_size=0.25,random_state=0)
    sc=StandardScaler()
    X_train=sc.fit_transform(X_train)
    X_test=sc.transform(X_test)
    return X_train, X_test,y_train,y_test

#Check how well the trained model predicts the test data.   
def cm_prediction(classifier,X_test):
    # Making the Confusion Matrix
    y_pred=classifier.predict(X_test)
    from sklearn.metrics import accuracy_score
    from sklearn.metrics import classification_report
    from sklearn.metrics import confusion_matrix
    cm = confusion_matrix(y_test, y_pred)
    Accuracy=accuracy_score(y_test,y_pred)
    report=classification_report(y_test,y_pred)
    return classifier,Accuracy,report,X_test,y_test,cm

#Used for classification problems, such as predicting whether a candidate will be selected or rejected.
def logistic(X_train,y_train,X_test):
    # Fitting K-NN to the Training set
    from sklearn.linear_model import LogisticRegression
    classifier=LogisticRegression(random_state=0)
    classifier.fit(X_train,y_train)
    classifier,Accuracy,report,X_test,y_test,cm=cm_prediction(classifier,X_test)
    return classifier,Accuracy,report,X_test,y_test,cm

#A linear SVM tries to separate classes using a straight decision boundary in the feature space.
def svm_linear(X_train,y_train,X_test):
    from sklearn.svm import SVC
    classifier=SVC(kernel='linear',random_state=0)
    classifier.fit(X_train,y_train)
    classifier,Accuracy,report,X_test,y_test,cm=cm_prediction(classifier,X_test)
    return classifier,Accuracy,report,X_test,y_test,cm

#The RBF kernel can learn more complex, curved decision boundaries.
#Use it when the classes cannot be separated well by a straight boundary.
def svm_NL(X_train,y_train,X_test):
    from sklearn.svm import SVC
    classifier=SVC(kernel='rbf',random_state=0)
    classifier.fit(X_train,y_train)
    classifier,Accuracy,report,X_test,y_test,cm=cm_prediction(classifier,X_test)
    return classifier,Accuracy,report,X_test,y_test,cm

#Naive Bayes uses probability to predict the class. Gaussian Naive Bayes assumes each feature follows a Gaussian 
#distribution within each class.
def Navie(X_train,y_train,X_test):       
        # Fitting K-NN to the Training set
        from sklearn.naive_bayes import GaussianNB
        classifier = GaussianNB()
        classifier.fit(X_train, y_train)
        classifier,Accuracy,report,X_test,y_test,cm=cm_prediction(classifier,X_test)
        return  classifier,Accuracy,report,X_test,y_test,cm         
    
#KNN looks at the 5 nearest training examples and predicts the class based on their labels.
#Here, p=2 means Euclidean distance.   
def knn(X_train,y_train,X_test):
           
        # Fitting K-NN to the Training set
        from sklearn.neighbors import KNeighborsClassifier
        classifier = KNeighborsClassifier(n_neighbors = 5, metric = 'minkowski', p = 2)
        classifier.fit(X_train, y_train)
        classifier,Accuracy,report,X_test,y_test,cm=cm_prediction(classifier,X_test)
        return  classifier,Accuracy,report,X_test,y_test,cm
    
#A Decision Tree makes a series of decisions, similar to a flowchart.
#criterion='entropy' controls how the tree evaluates splits.
def Decision(X_train,y_train,X_test):
        # Fitting K-NN to the Training set
        from sklearn.tree import DecisionTreeClassifier
        classifier = DecisionTreeClassifier(criterion = 'entropy', random_state = 0)
        classifier.fit(X_train, y_train)
        classifier,Accuracy,report,X_test,y_test,cm=cm_prediction(classifier,X_test)
        return  classifier,Accuracy,report,X_test,y_test,cm      

#Random Forest builds multiple decision trees and combines their predictions.
def random(X_train,y_train,X_test):
        
        # Fitting K-NN to the Training set
        from sklearn.ensemble import RandomForestClassifier
        classifier = RandomForestClassifier(n_estimators = 10, criterion = 'entropy', random_state = 0)
        classifier.fit(X_train, y_train)
        classifier,Accuracy,report,X_test,y_test,cm=cm_prediction(classifier,X_test)
        return  classifier,Accuracy,report,X_test,y_test,cm

#Store the accuracy results from the seven algorithms in one table.Suppose the input lists contain their accuracies.
def selectk_Classification(acclog,accsvml,accsvmnl,accknn,accnav,accdes,accrf): 
    
    dataframe=pd.DataFrame(index=['ChiSquare'],columns=['Logistic','SVMl','SVMnl','KNN','Navie','Decision','Random'])
    #Enumerate is like range which helps index positioning.
    for number,idex in enumerate(dataframe.index):      
        dataframe['Logistic'][idex]=acclog[number]       
        dataframe['SVMl'][idex]=accsvml[number]
        dataframe['SVMnl'][idex]=accsvmnl[number]
        dataframe['KNN'][idex]=accknn[number]
        dataframe['Navie'][idex]=accnav[number]
        dataframe['Decision'][idex]=accdes[number]
        dataframe['Random'][idex]=accrf[number]
    return dataframe
    
```

```python
#read dataset. index_col=None tells pandas not to use any column as the index automatically.
dataset1=pd.read_csv("prep.csv",index_col=None)

df2=dataset1
#Machine Learning algorithms generally require numerical input, 
#so this converts categorical columns into dummy (0/1) columns.
#drop_first=True removes one category to avoid redundant dummy columns.
df2 = pd.get_dummies(df2, drop_first=True)
#Removes the target/output column from the input features.
Indep_X=df2.drop('classification_yes', 1)
dep_Y=df2['classification_yes']
```

```python
#selecting the best no of(3) features and creating empty lists to store the accuracy of 7 classification algorithms.
kbest=selectkbest(Indep_X,dep_Y,3)       

acclog=[]
accsvml=[]
accsvmnl=[]
accknn=[]
accnav=[]
accdes=[]
accrf=[]
```

```python
#split train and test set
X_train, X_test, y_train, y_test=split_scalar(kbest,dep_Y)  
```

```python
#training 7 Machine Learning classification algorithms, calculating their accuracy, storing the accuracy values, 
#and comparing the results.
classifier,Accuracy,report,X_test,y_test,cm=logistic(X_train,y_train,X_test)
acclog.append(Accuracy)

classifier,Accuracy,report,X_test,y_test,cm=svm_linear(X_train,y_train,X_test)  
accsvml.append(Accuracy)

classifier,Accuracy,report,X_test,y_test,cm=svm_NL(X_train,y_train,X_test)  
accsvmnl.append(Accuracy)

classifier,Accuracy,report,X_test,y_test,cm=knn(X_train,y_train,X_test)  
accknn.append(Accuracy)
    
classifier,Accuracy,report,X_test,y_test,cm=Navie(X_train,y_train,X_test)  
accnav.append(Accuracy)
    
classifier,Accuracy,report,X_test,y_test,cm=Decision(X_train,y_train,X_test)  
accdes.append(Accuracy)
    
classifier,Accuracy,report,X_test,y_test,cm=random(X_train,y_train,X_test)  
accrf.append(Accuracy)
    
result=selectk_Classification(acclog,accsvml,accsvmnl,accknn,accnav,accdes,accrf)

```
